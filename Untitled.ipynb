```python
# feature Selection

df.columns
```

```python
print("Unique values in each column:\n")

for col in df.columns:
    print(col, ":", df[col].nunique())
```

```python
df["Log_Date"] = pd.to_datetime(df["Log_Date"])

df["Year"] = df["Log_Date"].dt.year
df["Month"] = df["Log_Date"].dt.month
df["Day"] = df["Log_Date"].dt.day
df["DayOfWeek"] = df["Log_Date"].dt.dayofweek
df["IsWeekend"] = (df["DayOfWeek"] >= 5).astype(int)
```

```python
df.head()
```

```python
df = df.drop(columns=["Log_ID", "User_ID", "Log_Date"])
```

```python
print(df.shape)
print(df.columns)
```

```python
X = df.drop("Calories_Burned", axis=1)
y = df["Calories_Burned"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX columns:")
print(X.columns.tolist())
```

```python
numeric_features = X.select_dtypes(include=["int64", "float64"]).columns
categorical_features = X.select_dtypes(include=["object"]).columns

print("Numerical Features:")
print(numeric_features.tolist())

print("\nCategorical Features:")
print(categorical_features.tolist())
```

```python
# data splitting

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# preprocessing
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ye preprocessor bna h abhi
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features)
    ]
)
```

```python
df.head()
```

```python
# ab yaha pr apply kr diya


# ab X_train and X_test me saari value numbers me hogi


X_train_processed = preprocessor.fit_transform(X_train)
```

```python
X_test_processed = preprocessor.transform(X_test)
```

```python
print("Before preprocessing:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\nAfter preprocessing:")
print("X_train_processed:", X_train_processed.shape)
print("X_test_processed :", X_test_processed.shape)
```

```python
#   training the data  using  linearRegression algorithm   model training
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train_processed, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred_lr = lr_model.predict(X_test_processed)

print("First 10 predictions:")
print(y_pred_lr[:10])
```

```python
# model evaluation
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression Results")
print("-------------------------")
print("MAE  :", mae_lr)
print("MSE  :", mse_lr)
print("RMSE :", rmse_lr)
print("R²   :", r2_lr)
```

```python
#   decision treee


from sklearn.tree import DecisionTreeRegressor

dt_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=8
)

dt_model.fit(X_train_processed, y_train)

print("Decision Tree model trained successfully!")
```

```python
y_pred_dt = dt_model.predict(X_test_processed)

print("First 10 predictions:")
print(y_pred_dt[:10])
```

```python
mae_dt = mean_absolute_error(y_test, y_pred_dt)
mse_dt = mean_squared_error(y_test, y_pred_dt)
rmse_dt = np.sqrt(mse_dt)
r2_dt = r2_score(y_test, y_pred_dt)

print("Decision Tree Results")
print("--------------------")
print("MAE  :", mae_dt)
print("MSE  :", mse_dt)
print("RMSE :", rmse_dt)
print("R²   :", r2_dt)
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Decision Tree"],
    "MAE": [mae_lr, mae_dt],
    "MSE": [mse_lr, mse_dt],
    "RMSE": [rmse_lr, rmse_dt],
    "R²": [r2_lr, r2_dt]
})

display(comparison)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully!")
```

```python
y_pred_rf = rf_model.predict(X_test_processed)

print("First 10 predictions:")
print(y_pred_rf[:10])
```

```python
# Random Forest Evaluation

mae_rf = mean_absolute_error(y_test, y_pred_rf)
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Results")
print("---------------------")
print("MAE  :", mae_rf)
print("MSE  :", mse_rf)
print("RMSE :", rmse_rf)
print("R²   :", r2_rf)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        mae_lr,
        mae_dt,
        mae_rf
    ],
    "MSE": [
        mse_lr,
        mse_dt,
        mse_rf
    ],
    "RMSE": [
        rmse_lr,
        rmse_dt,
        rmse_rf
    ],
    "R²": [
        r2_lr,
        r2_dt,
        r2_rf
    ]
})

display(comparison)
```
